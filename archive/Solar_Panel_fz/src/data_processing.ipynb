```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
site_1_data = pd.read_csv('../data/site_1_data.csv')
site_2_data = pd.read_csv('../data/site_2_data.csv')
site_3_data = pd.read_csv('../data/site_3_data.csv')
site_4_data = pd.read_csv('../data/site_4_data.csv')
```

```python
def remove_empty_gen(data):
    # Remove rows with solar_gen_kwh = 0
    data = data[data['solar_gen_kwh'] != 0]
    return data

def remove_outlier(data):
    # Remove rows with solar_gen_kwh > 1000
    data = data[data['solar_gen_kwh'] > 0.4]
    return data

def group_n_max(data, n):
    # Group by date and get the n max values
    group_data = data.groupby(data.index // 50).max()
    return group_data
```

```python
new_site_1_data = remove_empty_gen(site_1_data)
new_site_1_data = remove_outlier(new_site_1_data)
```

```python
# plot the data solar_gen_kwh vs time
plt.figure(figsize=(10, 6))
plt.plot(new_site_1_data['solar_gen_kwh'], label='Site 1')
plt.show()
```

```python
# plot the data solar_gen_kwh vs time
refined_site_1_data = group_n_max(new_site_1_data, 50)
plt.figure(figsize=(10, 6))
plt.plot(refined_site_1_data['solar_gen_kwh'], label='Site 1')
```
